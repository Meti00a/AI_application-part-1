den här koden tränar på en linjär regressionsmodell och sparar i en fil. Sedan läser den in modellen igen och används för att göra förutsägelser. Det är bra att man sparar modellen för att kunna användare senare istället för att träna den på nytt varje gång.

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from joblib import dump, load


X, y = make_regression( 
    n_samples=20000,
    n_features=3,
    noise=0.1  
)

model = LinearRegression().fit(X, y)

dump(model, "linear_model.joblib")

model_loaded = load("linear_model.joblib")


print(model_loaded.predict(X[:5]))
```
